# 1. Variables

A variable is a name that points to a value. Think of a labelled box: the label is the name, and the box holds the value.

In AI Engineering you would meet variables contantly just as i mentioned in the previous part. Every LLM request has setting, and each setting is a variable. For instance:

```python
# settings you woud pass to an LLM

model_name = "openai/gpt-oss-120b"          # -------> text 
temperature = 0.7                           # -------> decimal number: this shows how creative the model is in generation
max_tokens = 1024                           # -------> whole number: This revealed the longest reply we allow.
stream = False                              # -------> True or False: Show the reply as it is generated? 


print(model_name)
print(temperature)
print(max_tokens)
print(stream)
```

# What does this means?

**Line by line:**

- `model_name = "openai/gpt-oss-120b"` reads as *"store the text `openai/gpt-oss-120b` under the name `model_name`"*. The single `=` means **assign**, not "equals".
- Everything after `#` is a **comment**. Python ignores it; it's a note for humans. I spoke about this in the previous part.
- `print(...)` displays a value below the cell.

Variables can change (that's why they're called *variables*), for instance:

```python
Temperature = 0.2
print("New temperature:", temperature)
```

```python
# Update a variable using it's own old value
max_tokens = max_tokens + 512           # read 1024, add 512, store 1536 back into max_tokens
print("max_tokens is now", max_tokens)
```

### Naming rules

| Rule | Good | Bad |
|---|---|---|
| Letters, digits, underscores only | `max_tokens` | `max-tokens` |
| Cannot start with a digit | `model_2` | `2model` |
| No spaces | `user_name` | `user name` |
| Case-sensitive | `Temperature` and `temperature` are **different** variables | |
| Be descriptive | `total_cost` | `x`, `tc` |

Python style (called PEP 8) uses `snake_case` for variables: lowercase words joined with underscores.

# Constant

**Constants** are values that should never change while the program runs. By convention they use `ALL_CAPS`, that is capital letters:

```python
MAX_RETRIES = 3         # ---------> This is like, how many times we'd retry a failed API call
DEFAULT_MODEL = "openai/gpt-oss-120b"

print(MAX_RETRIES, DEFAULT_MODEL)
```
