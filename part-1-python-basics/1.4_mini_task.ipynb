# A mini "pre-flight check" for a prompt

Before real AI apps call a model, they run checks like these. We'll use a **flag variable** (`is_ready`): assume everything is fine, flip it to `False` whenever we find a problem, and only continue if it is still `True`.

Try changing `raw_input_text` and `temperature` to trigger each problem.

```python
raw_input_text = "  WHAT is     a Neural Network?   "
temperature = 0.7
max_tokens = 300


# 1. Clean the input (Keep the original capitalisation for the model)
cleaned_message = " ".join(raw_input_text.split())
check_text = cleaned_message.lower()

# 2. Run the checks
is_ready = True

if cleaned_message == "":
    print("Problem: the message is empty")
    is_ready = False

if "password" in check_text or "secret" in check_text:
    print("Proble: the message may contain sensitive information")
    is_ready = False


# 3. Only build the prompt if everything passed
if is_ready:
    prompt = f"""You are a helpful assistant.
Answer in simple language.

Question: {cleaned_message}"""

    estimated_tokens = len(prompt) // 4
    print(prompt)
    print("-" * 40)
    print(f"Estimated input tokens: ~{estimated_tokens}")
    print(f"Max reply tokens      : {max_tokens:,}")
    print(f"Temperature           : {temperature}")
else:
    print("Pre-flight check failed. Not sending to the model.")
```

## Common mistakes

| Mistake | What happens | Fix |
|---|---|---|
| `=` instead of `==` in an `if` | `SyntaxError` | `==` compares; `=` assigns |
| Missing `:` after `if` / `elif` / `else` | `SyntaxError` | add the colon |
| Wrong indentation | `IndentationError` | indent the body by 4 spaces |
| `"5" + 5` | `TypeError` | convert first: `int("5") + 5` |
| Forgetting to store a string method's result | the original doesn't change | `text = text.strip()` |
| Quotes around `True`/`None` | they become text, not booleans | write `True`, `None` unquoted |
| Comparing floats with `==` | surprising `False` | use `round()` or a tolerance |

## Recap

- **Variables** store values; use clear `snake_case` names and `ALL_CAPS` for constants.
- **Types** (`str`, `int`, `float`, `bool`, `None`) decide what operations work. Convert with `int()`, `float()`, `str()`.
- **Strings** are how prompts are built and cleaned: f-strings, slicing, `strip()`, `lower()`, `split()`, `join()`.
- **Conditionals** make decisions: `if` / `elif` / `else`, combined with `and` / `or` / `not`.
