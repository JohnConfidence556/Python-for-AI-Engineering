# Conditionals

A **conditional** lets your program make decisions: *"if this is true, do that; otherwise do something else."*

AI applications are full of them: Is the API key set? Is the temperature valid? Is the prompt too long? Did the model's reply contain what we need?

### `if` / `else`

```python
temperature = 1.5

if temperature > 1:
    print("Temperature is too high")

else:
    print("Temperature is fine")
```

**Two critical details:**

1. The line with `if` ends with a **colon `:`**
2. The code that belongs to it is **indented** (4 spaces). Indentation is how Python knows what is "inside" the `if`. Wrong indentation is usualy the number one mistake beginners make.

### Comparison operators

```python
print(5 == 5)       # equal to        (two = signs; one = is assignment!)
print(5 != 3)       # not equal to
print(5 > 3)        # greater than
print(5 < 3)        # less than
print(5 >= 5)       # greater than or equal
print(5 <= 4)       # less than or equal
print("a" == "A")   # False: text comparison is case-sensitive
```

### `elif`: more than two outcomes

Python checks from top to bottom and runs the **first** branch that is `True`, then skips the rest. Here's a simple routing rule that picks a model size by input length (a toy rule for practice; real routing considers task difficulty too):

```python
word_count = 180

if word_count > 500:
    model_tier = "large"

elif word_count > 100:
    model_tier = "small"

else:
    model_tier = "small"

print(f"{word_count} words -> use the {model_tier} model")
```

> **Order matters.** If we had tested `word_count > 100` first, a 600-word message would stop there and never reach the "large" branch. Test the most specific or strictest condition first.

### Combining conditions: `and`, `or`, `not`

| Operator | Meaning |
|---|---|
| `a and b` | both must be True |
| `a or b` | at least one must be True |
| `not a` | flips True/False |

```python
has_api_key = True
user_message = "Hello"
temperature = 0.7

if has_api_key and user_message:
    print("Ready to send")

if not has_api_key or not user_message:
    print("something is missing")

# Chained comparison: Python lets you write a range check naturally
if 0 <= temperature <= 1:
    print("Temperature is in the valid range")
```

### Checking for `None`: the missing API key

Programs read secrets such as API keys from **environment variables**. If the variable isn't set, Python gives you `None` instead of crashing. Always test for it with `is None`:

```python
import os   # os is a built-in module for talking to your operating system

api_key = os.getenv("ANTHROPIC_API_KEY")   # returns None if the variable isn't set

if api_key is None:
    print("No API key found. Copy .env.example to .env and add your key.")
else:
    print("API key found. (We never print the key itself!)")
```
