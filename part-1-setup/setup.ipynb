# 2. To know which python version the notebook is using

`sys` is a built-in module that describes the running interpreter. The path should contain `.venv`. If it doesn't, select the right kernel.

```python
import sys

print("python executable:", sys.executable)
print("Python version   :", sys.version.split()[0])
```

# 3. To check if you are in a virtual environment
Inside a venv, `sys.prefix` (the venv folder) differs from `sys.base_prefix` (the original Python install).

```python
in_venv = sys.prefix != sys.base_prefix
print("virtual environment active:", in_venv)
```

# 4. Understanding Variables and f-strings

A variable is a name that hold or store a value. And this values are reusable. Variables are a constant in the whole of writing python scripts even in AI Engineering.

An f-string lets you place a variable inside a print function. This is one form by which this variables can be reusable. The variables in an f-string is usually placed inside the curly bracket {}

```python
name = "future AI engineer"    # ------------> "name" here is a variable use to store the sting value "future AI engineer"

print(name)         # ------------------> Here's how you print the value of the variable "name"
```

" # " ---------------> This is an hashtag and it's used for commenting in python.

You write comment in a python script using "#" before the statement you want to write. 
comment allows your programs or script to be readable by yourself and other developers and they are not executable when you run a python program.

```python
model_id = "openai/gpt-oss-120b"

print(f"You are currently using the {model_id} model on groq")
```

Here my variable is "model_id" and i was able to use it in a print statement using f-strings.
